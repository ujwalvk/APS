```python
#1
import random
def prob_coin_toss(n):
    head=0
    tail=0
    for _ in range(n):
        event=random.choice(['Head', 'Tail'])
        if event=='Head':
            head =head+1
        else:
            tail=tail+1
    print('Number of  heads: ', head)
    print('Number of  Tail: ', tail)
    print('probability of getting a head: ', head/n)
    print('probability of getting a tail:  ', tail/n)
prob_coin_toss(100)
```

```python
#2. 
# 10 Tosses
print('For 10 Tosses: \n')
prob_coin_toss(10)
#100 Tosses
print('\nFor 100 Tosses: \n')
prob_coin_toss(100)
#1000 Tosses
print('\nFor 1000 Tosses: \n')
prob_coin_toss(1000)
#10000 tosses
print('\nFor 10000 Tosses: \n')
prob_coin_toss(10000)
```

```python
#4
n=10000
count_1=0
count_6=0
count1or6=0 
for _ in range(1,6):
    event = random.choice([1, 2, 3, 4, 5, 6])
    if event == 1:
        count_1 =  count_1+ 1
        count1or6 = count1or6 + 1
    elif event == 6:
        count_6 =  count_6+ 1
        count1or6 = count1or6 + 1

prob_1 = count_1 / n
prob_6 = count_6 / n
prob_1or6 = count1or6 / n
print('Probability of geeting 1  :', prob_1)
print('Probability of geeting 2  :', prob_6)
print('P(1) + P(6): ', prob_1 + prob_6)
print('P(A union B): ', prob_1or6)
if prob_1 + prob_6==prob_1or6:
    print('Disjoint set')
else:
    print('Not a Disjoint set')
```

```python
s={1,2,3,4,5,6}

a={2,4,6}
b={4,5,6}

union = a | b         
intersection = a & b 

P_a= len(a)/len(s)
P_b= len(b)/len(s)
P_union = len(union) / len(s)
P_intersection = len(intersection) / len(s)

print('P(A) =', P_a)
print('P(B) =', P_b)
print('P(A intersection B) =', P_intersection)
print('P(A union B) =', P_union)
print('P(A) + P(B) - P(A intersection B) =', P_a + P_b - P_intersection)
```

```python
#6
import random
n=10000
count_A = 0        
count_B = 0        
count_A_and_B = 0  

for _ in range(n):
    coin = random.choice(['Head', 'Tail'])
    die = random.choice([1, 2, 3, 4, 5, 6])

    event_A = coin == 'Head'
    event_B = die in [2, 4, 6]

    if event_A:
        count_A = count_A + 1
    if event_B:
        count_B = count_B + 1
    if event_A and event_B:
        count_A_and_B = count_A_and_B + 1

P_A = count_A / n
P_B = count_B / n
P_A_and_B = count_A_and_B / n

print('P(A) getting  Head: ', P_A)
print('P(B) - getting even number: ', P_B)
print('P(A) * P(B): ', P_A * P_B)
print('P(A intersection B): ', P_A_and_B)
```
